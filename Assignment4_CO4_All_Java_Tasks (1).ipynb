## Task 1 – Student Fee Payment

```python
%%writefile FeePayment.java
import java.util.Scanner;

public class FeePayment {
    static class FeeService {
        private String receipt = "No payment made.";

        public synchronized String pay(String regNo, double amount) {
            receipt = "Receipt: " + regNo + " | Amount: Rs. " + amount;
            return "Payment Successful!";
        }

        public synchronized String getReceipt() {
            return receipt;
        }
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        FeeService service = new FeeService();

        while (true) {
            System.out.println("\\n1. Pay Fee  2. View Receipt  3. Exit");
            System.out.print("Choose: ");
            int choice = sc.nextInt();
            sc.nextLine();

            if (choice == 1) {
                System.out.print("Register Number: ");
                String regNo = sc.nextLine().trim();
                System.out.print("Amount: ");
                double amount = sc.nextDouble();
                sc.nextLine();

                if (regNo.isEmpty() || amount <= 0) {
                    System.out.println("Enter valid details.");
                } else {
                    Thread paymentThread = new Thread(() ->
                        System.out.println(service.pay(regNo, amount)));
                    paymentThread.start();
                    try {
                        paymentThread.join();
                    } catch (InterruptedException e) {
                        Thread.currentThread().interrupt();
                    }
                }
            } else if (choice == 2) {
                System.out.println(service.getReceipt());
            } else if (choice == 3) {
                break;
            } else {
                System.out.println("Invalid choice.");
            }
        }
        sc.close();
    }
}
```

## Task 2 – Hospital Appointment Desk

```python
%%writefile HospitalAppointment.java
import java.util.*;

public class HospitalAppointment {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        Map<String, List<String>> doctors = new LinkedHashMap<>();
        doctors.put("Cardiology", Arrays.asList("Dr. Rao", "Dr. Mehta"));
        doctors.put("Neurology", Arrays.asList("Dr. Kumar", "Dr. Sharma"));
        doctors.put("Orthopedics", Arrays.asList("Dr. Reddy", "Dr. Das"));

        String[] slots = {"09:00 AM", "10:00 AM", "11:00 AM", "02:00 PM"};

        System.out.print("Patient ID: ");
        String patientId = sc.nextLine().trim();

        if (patientId.isEmpty()) {
            System.out.println("Patient ID is required.");
            sc.close();
            return;
        }

        List<String> departments = new ArrayList<>(doctors.keySet());
        System.out.println("\\nDepartments:");
        for (int i = 0; i < departments.size(); i++)
            System.out.println((i + 1) + ". " + departments.get(i));

        System.out.print("Choose department: ");
        int d = sc.nextInt();
        sc.nextLine();

        if (d < 1 || d > departments.size()) {
            System.out.println("Invalid department.");
            sc.close();
            return;
        }

        String department = departments.get(d - 1);
        List<String> doctorList = doctors.get(department);

        System.out.println("\\nDoctors:");
        for (int i = 0; i < doctorList.size(); i++)
            System.out.println((i + 1) + ". " + doctorList.get(i));

        System.out.print("Choose doctor: ");
        int doc = sc.nextInt();

        System.out.println("\\nTime Slots:");
        for (int i = 0; i < slots.length; i++)
            System.out.println((i + 1) + ". " + slots[i]);

        System.out.print("Choose time slot: ");
        int time = sc.nextInt();

        if (doc < 1 || doc > doctorList.size() || time < 1 || time > slots.length) {
            System.out.println("Invalid selection.");
        } else {
            System.out.println("\\nAppointment Booked Successfully!");
            System.out.println("Patient ID: " + patientId);
            System.out.println("Department: " + department);
            System.out.println("Doctor: " + doctorList.get(doc - 1));
            System.out.println("Time: " + slots[time - 1]);
        }
        sc.close();
    }
}
```

## Task 3 – Campus Library Search

```python
%%writefile LibrarySearch.java
import java.util.*;

public class LibrarySearch {
    static class Book {
        String title, author, subject;

        Book(String title, String author, String subject) {
            this.title = title;
            this.author = author;
            this.subject = subject;
        }
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        List<Book> books = Arrays.asList(
            new Book("Java Programming", "Herbert Schildt", "Programming"),
            new Book("Effective Java", "Joshua Bloch", "Programming"),
            new Book("Data Structures", "Seymour Lipschutz", "Computer Science"),
            new Book("Database Management Systems", "Raghu Ramakrishnan", "Database")
        );

        System.out.print("Enter search text: ");
        String query = sc.nextLine().trim().toLowerCase();

        System.out.println("Search by: 1. All  2. Book  3. Author  4. Subject");
        System.out.print("Choose category: ");
        int category = sc.nextInt();

        boolean found = false;
        for (Book book : books) {
            boolean match;
            if (category == 2)
                match = book.title.toLowerCase().contains(query);
            else if (category == 3)
                match = book.author.toLowerCase().contains(query);
            else if (category == 4)
                match = book.subject.toLowerCase().contains(query);
            else
                match = book.title.toLowerCase().contains(query)
                     || book.author.toLowerCase().contains(query)
                     || book.subject.toLowerCase().contains(query);

            if (match) {
                System.out.println("\\nBook: " + book.title);
                System.out.println("Author: " + book.author);
                System.out.println("Subject: " + book.subject);
                found = true;
            }
        }

        if (!found)
            System.out.println("No matching books found.");

        System.out.print("\\nEnter a book title to reserve (or press Enter to skip): ");
        sc.nextLine();
        String reserve = sc.nextLine().trim();
        if (!reserve.isEmpty())
            System.out.println("Reserved: " + reserve);

        sc.close();
    }
}
```

## Task 4 – Event Handling Forensics / Drawing Pad

```python
%%writefile DrawingPad.java
import java.util.*;

public class DrawingPad {
    static class Point {
        int x, y;
        Point(int x, int y) {
            this.x = x;
            this.y = y;
        }
    }

    interface MouseClickListener {
        void mouseClicked(Point point);
    }

    static class DrawingArea {
        List<Point> points = new ArrayList<>();
        int clickCount = 0;

        MouseClickListener listener = new MouseClickListener() {
            public void mouseClicked(Point point) {
                points.add(point);
                clickCount++;
                System.out.println("Mouse click " + clickCount +
                                   " recorded at (" + point.x + ", " + point.y + ")");
            }
        };

        void simulateClick(int x, int y) {
            listener.mouseClicked(new Point(x, y));
        }

        void displayPoints() {
            System.out.println("\\nDrawing points:");
            for (Point p : points)
                System.out.println("(" + p.x + ", " + p.y + ")");
            System.out.println("Total clicks: " + clickCount);
        }
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        DrawingArea area = new DrawingArea();

        System.out.println("Drawing Pad - Mouse Click Simulation");
        System.out.println("Enter x and y coordinates. Enter -1 for x to finish.");

        while (true) {
            System.out.print("X coordinate: ");
            int x = sc.nextInt();
            if (x == -1)
                break;

            System.out.print("Y coordinate: ");
            int y = sc.nextInt();
            area.simulateClick(x, y);
        }

        area.displayPoints();
        sc.close();
    }
}
```

## Task 5 – Currency Converter: Four Event Handling Styles

```python
%%writefile CurrencyConverter.java
import java.util.Scanner;

public class CurrencyConverter {
    interface ActionListener {
        void actionPerformed(double amount, String currency);
    }

    static double rate(String currency) {
        return currency.equals("USD") ? 0.012 : 0.011;
    }

    static void convert(double amount, String currency) {
        if (amount < 0) {
            System.out.println("Invalid amount.");
            return;
        }
        double converted = amount * rate(currency);
        System.out.printf("%.2f INR = %.2f %s%n", amount, converted, currency);
    }

    static class ImplementsListener implements ActionListener {
        public void actionPerformed(double amount, String currency) {
            convert(amount, currency);
        }
    }

    static class NamedInnerClass {
        class Handler implements ActionListener {
            public void actionPerformed(double amount, String currency) {
                convert(amount, currency);
            }
        }
    }

    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        System.out.println("Choose Event Handling Style:");
        System.out.println("1. Implements ActionListener");
        System.out.println("2. Named Inner Class");
        System.out.println("3. Anonymous Inner Class");
        System.out.println("4. Lambda Expression");
        System.out.print("Choice: ");
        int choice = sc.nextInt();

        System.out.print("Amount in INR: ");
        double amount = sc.nextDouble();

        System.out.print("Target currency (USD/EUR): ");
        String currency = sc.next().toUpperCase();

        if (!currency.equals("USD") && !currency.equals("EUR")) {
            System.out.println("Choose USD or EUR.");
            sc.close();
            return;
        }

        ActionListener listener = null;

        if (choice == 1) {
            listener = new ImplementsListener();
        } else if (choice == 2) {
            NamedInnerClass outer = new NamedInnerClass();
            listener = outer.new Handler();
        } else if (choice == 3) {
            listener = new ActionListener() {
                public void actionPerformed(double value, String target) {
                    convert(value, target);
                }
            };
        } else if (choice == 4) {
            listener = (value, target) -> convert(value, target);
        }

        if (listener == null)
            System.out.println("Invalid choice.");
        else
            listener.actionPerformed(amount, currency);

        sc.close();
    }
}
```

## Task 6 – Campus Event Registration Using JDBC (SQLite)

```python
%%writefile EventRegistration.java
import java.sql.*;
import java.util.*;
import java.io.*;

public class EventRegistration {
    static final String URL = "jdbc:sqlite:campus_events.db";

    static Connection connect() throws SQLException {
        return DriverManager.getConnection(URL);
    }

    static void createTable() throws SQLException {
        String sql = "CREATE TABLE IF NOT EXISTS registrations (" +
                     "id INTEGER PRIMARY KEY AUTOINCREMENT, " +
                     "student_id TEXT NOT NULL, " +
                     "student_name TEXT NOT NULL, " +
                     "event_name TEXT NOT NULL, " +
                     "registered_at DATETIME DEFAULT CURRENT_TIMESTAMP, " +
                     "UNIQUE(student_id, event_name))";
        try (Connection con = connect(); Statement st = con.createStatement()) {
            st.execute(sql);
        }
    }

    static void registerStudent(Scanner sc) {
        try {
            System.out.print("Student ID: ");
            String id = sc.nextLine().trim();
            System.out.print("Student Name: ");
            String name = sc.nextLine().trim();

            System.out.println("1. Cultural Fest  2. Sports Meet  3. Tech Symposium");
            System.out.print("Choose event: ");
            int choice = Integer.parseInt(sc.nextLine());

            String[] events = {"Cultural Fest", "Sports Meet", "Tech Symposium"};
            if (id.isEmpty() || name.isEmpty() || choice < 1 || choice > 3) {
                System.out.println("Enter valid details.");
                return;
            }

            String sql = "INSERT INTO registrations(student_id, student_name, event_name) VALUES(?,?,?)";
            try (Connection con = connect(); PreparedStatement ps = con.prepareStatement(sql)) {
                ps.setString(1, id);
                ps.setString(2, name);
                ps.setString(3, events[choice - 1]);
                ps.executeUpdate();
                System.out.println("Registration Successful!");
            }
        } catch (SQLException e) {
            System.out.println("Registration failed. Student may already be registered for this event.");
        } catch (NumberFormatException e) {
            System.out.println("Invalid event selection.");
        }
    }

    static void loadData(Scanner sc) {
        System.out.println("Filter: 1. All  2. Cultural Fest  3. Sports Meet  4. Tech Symposium");
        System.out.print("Choose filter: ");
        int choice;
        try {
            choice = Integer.parseInt(sc.nextLine());
        } catch (NumberFormatException e) {
            System.out.println("Invalid filter.");
            return;
        }

        String sql = "SELECT * FROM registrations";
        if (choice >= 2 && choice <= 4)
            sql += " WHERE event_name = ?";
        sql += " ORDER BY registered_at DESC";

        try (Connection con = connect(); PreparedStatement ps = con.prepareStatement(sql)) {
            if (choice >= 2 && choice <= 4) {
                String[] events = {"Cultural Fest", "Sports Meet", "Tech Symposium"};
                ps.setString(1, events[choice - 2]);
            }
            try (ResultSet rs = ps.executeQuery()) {
                System.out.println("\\nID | Student ID | Student Name | Event | Registered At");
                while (rs.next()) {
                    System.out.println(rs.getInt("id") + " | " +
                        rs.getString("student_id") + " | " +
                        rs.getString("student_name") + " | " +
                        rs.getString("event_name") + " | " +
                        rs.getString("registered_at"));
                }
            }
        } catch (SQLException e) {
            System.out.println("Database error: " + e.getMessage());
        }
    }

    static void exportCSV() {
        String sql = "SELECT * FROM registrations";
        try (Connection con = connect();
             Statement st = con.createStatement();
             ResultSet rs = st.executeQuery(sql);
             PrintWriter out = new PrintWriter(new FileWriter("registrations.csv"))) {

            out.println("ID,Student ID,Student Name,Event,Registered At");
            while (rs.next()) {
                out.println(rs.getInt("id") + "," +
                    quote(rs.getString("student_id")) + "," +
                    quote(rs.getString("student_name")) + "," +
                    quote(rs.getString("event_name")) + "," +
                    quote(rs.getString("registered_at")));
            }
            System.out.println("CSV exported to registrations.csv");
        } catch (Exception e) {
            System.out.println("Export failed: " + e.getMessage());
        }
    }

    static String quote(String value) {
        return "\\"" + value.replace("\\"", "\\"\\"") + "\\"";
    }

    public static void main(String[] args) throws Exception {
        createTable();
        Scanner sc = new Scanner(System.in);

        while (true) {
            System.out.println("\\nCAMPUS EVENT REGISTRATION");
            System.out.println("1. Register Student");
            System.out.println("2. View / Filter Registrations");
            System.out.println("3. Export CSV");
            System.out.println("4. Exit");
            System.out.print("Choose: ");
            String choice = sc.nextLine();

            switch (choice) {
                case "1": registerStudent(sc); break;
                case "2": loadData(sc); break;
                case "3": exportCSV(); break;
                case "4": sc.close(); return;
                default: System.out.println("Invalid choice.");
            }
        }
    }
}
```
